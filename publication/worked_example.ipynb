# Detailed Generation

```python
from alparc import *

# 1. Phonemes
phonemes = load_phonemes(lang="deu")
phonemes
```

```python

# 2. Syllables
syllables = make_syllables(phonemes)
syllables
```

```python

# 3. Words
words = make_words(
    syllables,
    n_syllables=3,
    n_words=100,
)
words
```

```python

# 4. Lexicons
lexicons = make_lexicons(
    words,
    n_lexicons=2,
    n_words=3,
)
lexicons
```

```python
streams = make_streams(
    lexicons,
    n_repetitions=1,
)

for stream in streams:
    print_stream(stream)
```

# Conditional Generation

```python
lexicons = [
    ['n_o|n_i|x_e', 'b_u|l_o|t_e', 'r_e|m_o|f_u', 'k_o|t_u|s_a'], 
    ['m_i|l_o|d_e', 'd_a|l_e|b_u', 'n_o|r_u|p_a', 'k_a|t_e|x_i'], 
]

lexicons = [diagnose(lexicon) for lexicon in lexicons]

streams = make_streams(
    lexicons,
    n_repetitions=1,
)

for stream in streams:
    print_stream(stream)
```
